```python
import torch
import numpy as np
```

1.直接创建

```python
t1 = torch.tensor([1, 2, 3])  # 1D tensor
t2 = torch.tensor([[1, 2], [3, 4]])  # 2D tensor
t3 = torch.tensor([1.5, 2.0], dtype=torch.float32)  # 指定数据类型（float32）
t4 = torch.tensor((1, 2, 3), dtype=torch.int64)  # 从元组创建
print(t1,t2,t3,t4)
```

2.从np数组创建

```python
np_arr = np.array([[1, 2], [3, 4]], dtype=np.float32)
t = torch.from_numpy(np_arr)
print(t)
```

3.创建指定形状的未初始化

```python
t = torch.empty((2, 3))  # 2行3列的未初始化tensor
print(t)
t = torch.empty(1, 2, 3, dtype=torch.int32)  # 3D tensor，指定dtype
print(t)
t=torch.Tensor(2,3)
print(t)
```

4.创建全0或者全1

```python
t=torch.zeros(3,4)
print(t)
```

5.创建序列

```python
t = torch.arange(5)
print(t)
t = torch.arange(1, 10, 2)
print(t)
```

```python
t = torch.linspace(0, 1, 5)
print(t)
```

```python
t = torch.logspace(0, 6, 3,base=2)
print(t)
```

6.随机初始化tensor

```python
t = torch.rand(2, 3)  # 2x3 矩阵，元素在[0,1)随机
print(t)
t = torch.randint(1, 10, (3,2))  # [1,10) 随机整数，形状(3,2)
print(t)
t = torch.randn(2, 2)  # 2x2 矩阵，元素服从N(0,1)
print(t)
t = torch.normal(mean=torch.tensor([0.0, 1.0]), std=torch.tensor([0.1, 0.2]))
print(t)
t = torch.randperm(5)  # 例如 [3, 0, 2, 4, 1] 随机采样
print(t)
```

7.特殊矩阵

```python
t = torch.eye(3)  # 3x3 单位矩阵（对角线为1，其余为0）
print(t)
t = torch.eye(2, 3)  # 2x3 矩阵，对角线为1
print(t)
t = torch.diag(torch.tensor([1, 2, 3]))  # 3x3 对角矩阵，对角线为1,2,3
print(t)
```

8.根据已有tensor创建

```python
t = torch.Tensor([[1, 2], [3, 4]])
t_zero = torch.zeros_like(t)  # 2x2 全零矩阵，dtype与t一致
t_one=torch.ones_like(t)
t_rand=torch.rand_like(t)
print(t)
print(t_zero)
print(t_one)
print(t_rand)
```
