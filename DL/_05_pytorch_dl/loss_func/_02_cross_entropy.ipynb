```python
import torch
import torch.nn as nn
```

1.目标值为顺序编码的标签

```python
input=torch.randn(6,8)
print(input)
#pred=torch.softmax(input,dim=1) 可以不转化 底层会自动处理
```

```python
#目标值
target=torch.tensor([1,0,3,7,5,2])
print(target)
```

```python
#定义损失函数
loss=nn.CrossEntropyLoss()
output=loss(input,target)
print(output)
```

2.目标值为独热编码

```python
input=torch.randn(6,8)
print(input)
target=torch.randn(6,8).softmax(dim=1)
print(target)
```

```python
loss=nn.CrossEntropyLoss()
output=loss(input,target)
print(output)
```
