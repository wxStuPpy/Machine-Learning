```python
import torch
```

```python
x=torch.tensor(2.0,requires_grad=True)
y=x.detach()
print(x,x.requires_grad,id(x))
print(y,y.requires_grad,id(y))
```

```python
print(x.untyped_storage().data_ptr())
print(y.untyped_storage().data_ptr())
```

```python
x=torch.ones((2,2),requires_grad=True)
y=x*x
print(x)
print(y)
```

```python
,#分离y
u=y.detach()
print(u)
```

u已经分离 相当于常数

```python
#让u参与新张量的计算
z=u*x
print(z)
```

```python
z.sum().backward()
print(x.grad)
```
