```python
import torch
```

用两组数据对比 detach和data的区别

```python
x1 = torch.tensor([1.0, 2, 3], requires_grad=True)
x2 = torch.tensor([1.0, 2, 3], requires_grad=True)
print(x1)
print(x2)
```

```python
y1=x1.sigmoid()
y2=x2.sigmoid()
print(y1)
print(y2)
```

```python
# y1.sum().backward()
# y2.sum().backward()
# print(x1.grad)
# print(x2.grad)
```

data vs detach
---
data修改不会被微分引擎跟踪 最好是只读 脱离了计算图
detach保证脱离后 不再进行反向传播 安全脱离计算图

```python
z1=y1.data
z2=y2.detach()
print(z1)
print(z2)
```

```python
z1.zero_()
z2.zero_()
print(y1)
print(y2)
```

```python
y1.sum().backward()
print(x1.grad)
```

已经脱离了计算图 不能进行反向传播
