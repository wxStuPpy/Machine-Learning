1.张量元素类型转化

```python
tensor1=torch.tensor([1,2,3])
print(tensor1.dtype)
```

```python
tensor1=tensor1.type(torch.float)
print(tensor1.dtype)
```

```python
tensor1=tensor1.half()
print(tensor1.dtype)
```

```python
tensor1=tensor1.to(torch.complex64)
print(tensor1.dtype)
print(tensor1)
```

2.tensor与nd.array相互转化
---
1.tensor-->ndarray

```python
tensor1=torch.rand(3,2)
np.set_printoptions(precision=4)
print(tensor1)
```

```python
ndarray1=tensor1.numpy()
print(ndarray1)
```

```python
tensor1[1][1]=1.0
print(tensor1)
print(ndarray1)
```

发现会共享同一份内存 使用copy来避免

```python
ndarray2=tensor1.numpy().copy()
print(ndarray2)
tensor1[1][1]=2.1212
print(tensor1)
print(ndarray2)
```

2.ndarray-->torch 也是共享内存

```python
ndarray1=np.random.randn(3,2)
tensor1=torch.from_numpy(ndarray1)
print(tensor1)
print(ndarray1)
```

```python
tensor2=torch.from_numpy(ndarray2.copy())
tensor2[:,1]=tensor2[:,1]*2
print(tensor2)
print(ndarray2)
```
